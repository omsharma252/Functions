# Functions


1. What is the difference between a function and a method in Python?
 - A function is an independent block of code defined using def.
A method is a function defined inside a class and tied to an object.
Methods automatically take the instance (self) or class (cls) as the first argument.
Functions are called standalone; methods are called on objects.

2. Explain the concept of function arguments and parameters in Python.
 - Parameters are variables listed in a function's definition.
Arguments are the actual values passed to the function.
Parameters act as placeholders for the input data.
Arguments fill those placeholders during a function call.

3. What are the different ways to define and call a function in Python?
 - In Python, you can define a function using def with or without parameters, then call it by its name.
You can also use keyword arguments to call functions with named parameters for clarity.
Lambda functions are anonymous, single-expression functions defined with lambda.
All these methods allow flexible ways to define and call functions depending on the situation.
4. What is the purpose of the 'return' statement in a Python function?
 - The return statement in a Python function sends a value back to the caller.
It ends the function execution and provides the result that can be used elsewhere in the code.
Without return, a function returns None by default.
It allows functions to produce output instead of just performing actions.
5. What are iterators in Python and how do they differ from iterables?
 - Iterables are objects you can loop over (like lists, tuples, or strings).
Iterators are objects that represent a stream of data and produce the next item when you call next() on them.
All iterators are iterables, but not all iterables are iterators until you get an iterator from them using iter().
In short, iterables can be turned into iterators, which control the actual iteration process.
6.  Explain the concept of generators in Python and how they are defined.
 - Generators in Python are special functions that **yield** values one at a time, producing items lazily instead of returning them all at once.
They are defined like regular functions but use the 'yield' keyword to return values.
Each time 'yield' is called, the generator’s state is saved, allowing it to resume where it left off on the next call.
Generators are memory-efficient for handling large data sequences because they generate values on demand.
7.  What are the advantages of using generators over regular functions?
 - Generators offer several advantages over regular functions:

 - **Memory efficiency:** They yield items one at a time, so they don’t store the entire sequence in memory.
  - **Lazy evaluation:** Values are generated only when needed, improving performance with large datasets.
 - **Maintain state:** Generators automatically remember their position between yields without extra code.
 - **Simpler code:** They provide an easy way to implement iterators without writing classes.
8.  What is a lambda function in Python and when is it typically used?
 - A lambda function in Python is a small, anonymous function defined with the lambda keyword.
It can have any number of arguments but only one expression, which is returned automatically.
Lambda functions are typically used for short, simple operations, often as arguments to functions like map(), filter(), or sorted().
They help write concise, inline functions without formally defining a def block.
9. Explain the purpose and usage of the 'map()' function in Python.
  - The `map()` function in Python applies a given function to every item of an iterable (like a list) and returns an iterator with the results.

It’s useful for transforming or processing data without writing explicit loops.

Example usage:

```python
numbers = [1, 2, 3, 4]
squared = map(lambda x: x**2, numbers)
print(list(squared))  # Output: [1, 4, 9, 16]
```

This applies the lambda function to each number, returning their squares efficiently.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
 - Here’s a quick breakdown of the differences:

* **`map()`** applies a function to *every* item in an iterable and returns an iterator with the results.
* **`filter()`** applies a function that returns `True` or `False` to each item and returns only those items where the function is `True`.
* **`reduce()`** (from `functools`) applies a function cumulatively to items, reducing the iterable to a single value (like summing or multiplying all elements).

Example:

```python
from functools import reduce

nums = [1, 2, 3, 4]

# map: square each number
print(list(map(lambda x: x**2, nums)))  # [1, 4, 9, 16]

# filter: keep only even numbers
print(list(filter(lambda x: x % 2 == 0, nums)))  # [2, 4]

# reduce: multiply all numbers together
print(reduce(lambda x, y: x * y, nums))  # 24
```

Each serves a different purpose: transformation (`map`), selection (`filter`), and aggregation (`reduce`).

````python
# Create a Python function that accepts a string and returns the reverse of that string.
def reverse_string(input_string):
  reversed_string = input_string[::-1]
  return reversed_string

# Example usage:
my_string = "Hello, World!"
result = reverse_string(my_string)
print(result) # Print the result of the function call
````

````python
#Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

# User input and result
number = int(input("Enter the number to check: "))
if is_prime(number):
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
````

````python
# Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms
def fibonacci_generator(max_terms):
    a=0
    b=1
    for _ in range(max_terms):
        yield a
        a, b = b, a + b

# Example usage:
n = int(input("Enter the number of Fibonacci terms to generate: "))
for num in fibonacci_generator(n):
    print(num)
````

````python
#Write a generator function in Python that yields the powers of 2 up to a given exponent.
# Generator function to yield powers of 2
def powers_of_two(max_exponent):
    for exponent in range(max_exponent + 1):
        yield 2 ** exponent

# Get input from the user
n = int(input("Enter the maximum exponent: "))

# Print powers of 2 from 2^0 to 2^n
print(f"Powers of 2 from 2^0 to 2^{n}:")
for power in powers_of_two(n):
  print(power)
````

````python
#Implement a generator function that reads a file line by line and yields each line as a string.
def create_sample_file(filename):
    with open(filename, 'w') as f:
        f.write("Hello, world!\n")
        f.write("This is line 2.\n")
        f.write("And this is line 3.\n")

def read_file_line_by_line(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.rstrip('\n')

def main():
    filename = 'example.txt'
    create_sample_file(filename)
    for line in read_file_line_by_line(filename):
        print(line)

main()
````

````python
#Create a Python program that uses `filter()` to remove all the vowels from a given string.
def remove_vowels(input_string):
    vowels = 'aeiouAEIOU'

    filtered_chars = filter(lambda char: char not in vowels, input_string)

    return ''.join(filtered_chars)


text = "Hello, World!"
result = remove_vowels(text)
print(result)
````
